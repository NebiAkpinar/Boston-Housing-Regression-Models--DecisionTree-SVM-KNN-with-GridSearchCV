```python
from sklearn.model_selection import train_test_split
```

```python
X=df.drop("MEDV",axis=1)
y=df["MEDV"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=15)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
sns.boxenplot(X_train_scaled)
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import GridSearchCV 
```

```python
param_grid={
    "criterion":['squared_error','absolute_error','friedman_mse'],
    "splitter":["best", "random"],
    "max_depth":[1,2,3,4,5,15,None]
}
```

```python
tree_reg=DecisionTreeRegressor() #default olanları denedim GridSearchCv ile diğerlerini denicem
```

```python
tree_reg.fit(X_train,y_train)
```

```python
y_pred=tree_reg.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
print("R2_Score",r2_score(y_pred,y_test))
print("MAE",mean_absolute_error(y_pred,y_test))
print("MSE",mean_squared_error(y_pred,y_test))
```

```python
grid=GridSearchCV(estimator=DecisionTreeRegressor(),param_grid=param_grid,n_jobs=-1)
```

```python
import warnings
warnings.filterwarnings("ignore")
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred=grid.predict(X_test)
```

```python
print("R2_Score", r2_score(y_test, y_pred))
print("MAE", mean_absolute_error(y_test, y_pred))
print("MSE", mean_squared_error(y_test, y_pred))
```
